# Dado un número cualquiera decir si es polidivisible. 

```python
numero = 381654729
numero_str = str(numero)
contador = 0

polidivisible = True
while contador < len(numero_str):
    
    numero_cortado = numero_str[:contador +1]
    numero_cortado_int = int(numero_cortado)
    
    if numero_cortado_int % (contador + 1) != 0:
        polidivisible = False    
    contador += 1
    
    
if polidivisible:
    print("El número es polidivisible")
    
else:
    print("No es polidivisible")
```

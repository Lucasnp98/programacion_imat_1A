### Solicitar info de futbolistas. La clave es el nombre, el valor es una lita compuesta por:
- Equipo (String), Edad (int), Patrocinador (String), Goles-tarjetas en tupla. 

```python
num_jugadores = 3
jugadores = {}

contador = 0
while contador < num_jugadores:
    
    nombre = input("Nombre: ")
    
    equipo = input("Equipo: ")
     
    pat = input("Patrocinador: ")
    edad = int(input("Edad: "))
     
    goles = int(input("Goles: "))
    tarjetas = int(input("Tarjetas: "))
     
    tupla_goles_tarjetas = (goles, tarjetas)
    
    info_futbolista = [equipo, pat, edad, tupla_goles_tarjetas]
    
    
    # jugadores.append() ESTO NO FUNCIONARÍA YA QUE APPEND ES UN MÉTODO DE LAS LISTAS. NECESITAMOS EL "APPEND" DE 
    # LOS DICCIONARIOS
    
    jugadores[nombre] = info_futbolista
    
    
    
    contador += 1


print(jugadores)
```

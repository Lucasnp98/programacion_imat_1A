```python
coches = [
    ["Audi A3", "Negro", 150000, 1350],
    ["BMW Serie 3", "Blanco", 85000, 1550],
    ["Seat León", "Rojo", 120000, 1280],
    ["Volkswagen Golf", "Gris", 65000, 1320],
    ["Toyota Corolla", "Azul", 40000, 1400],
    ["Mercedes Clase A", "Negro", 95000, 1450],
    ["Renault Clio", "Blanco", 175000, 1100],
    ["Ford Focus", "Rojo", 110000, 1300],
    ["Hyundai Tucson", "Gris", 30000, 1600],
    ["Peugeot 308", "Azul", 78000, 1250]
]


umbral_km_inf = 3000
umbral_km_superior = 20000
# Cuenta cuantos coches están dentro de ese umbral de km, obten una lista con 
# los modelos de esos coches y obten otra lista con los modelos de los coches que no lo cumplan
# Ahora recorres con el while!
modelos_coches_no_cumplen = []
posicion = 0
modelos_coches_cumplen = []
while posicion < len(coches):
    lista_coche = coches[posicion]
    modelo = lista_coche[0]
    km_coche = lista_coche[2]
    # modelo = coches[posicion][0]
    # km_coche = coches[posicion][2]
    
    if km_coche > umbral_km_inf and km_coche < umbral_km_superior:
        modelos_coches_cumplen.append(modelo)
    else:
        modelos_coches_no_cumplen.append(modelo)
        
    
    
    
    posicion += 1
```

```python
es_primo = True
divisor = 2
numero = 9

#Falta checkear el número

while divisor < numero:
    resto = numero % divisor

    if resto == 0:
        es_primo = False
        
    divisor = divisor + 1
    
if es_primo:
    print("El número: ", numero, "es primo!")
else:
    print("El número", numero, "no es primo")


```

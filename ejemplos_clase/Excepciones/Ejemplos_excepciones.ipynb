```python
class DNIError(Exception):
    pass


```

```python
def checkear_dni(dni:str, letras:str):
    
    
    if len(dni) != 9:
        raise DNIError("La longiitud es distinta de 9")
        
    numero_dni = dni[:-1]
    
    if not numero_dni.isdecimal():
        raise DNIError("Ojo que el numero del dni no es un numero")
    
    numero_dni = int(numero_dni)
    letra_dni = dni[-1]
    if not letra_dni.isalpha():
        raise DNIError("La letra el dni no es letra!!!!")
    
    resto = numero_dni % 23
    letra_correcta = letras[resto]
    if letra_dni != letra_correcta:
        raise DNIError(f"La letra {letra_dni} no es correcta, la correcta es: {letra_correcta}")
    
    
    
    
```

```python
letras = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
dni_prueba = "123456799"
try:
    checkear_dni(dni_prueba, letras)
except DNIError as error:
    print(error)
except ValueError as error:
    print(error)


print("Mi programa sigue")
```
